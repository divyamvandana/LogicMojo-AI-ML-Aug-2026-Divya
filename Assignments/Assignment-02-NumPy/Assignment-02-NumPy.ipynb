## Problem: 7 
Create a function which creates an $n×n$ array with $(i , j)$ -entry equal to $i+ j$

```python
np.arange(5).reshape(-1, 1) + [0,1,2,3,4]

# np.arange(5) -- array([0, 1, 2, 3, 4, 5])
# np.arange(5)[:, None]
# np.arange(5).reshape(-1, 1)
```

```python
n = 7
m = np.ones((n, n), dtype=int)
# Set the first row to 0, 1, ..., n-1
m[0, :] = np.arange(n)
np.cumsum(m, axis=0)
```

```python
def create_sum_matrix(n):
    # Create an n x n array using broadcasting
    #np.arange(5).reshape(-1, 1) 
    return np.arange(n)[:, None] + np.arange(n)


print('Using broadcasting')
print(create_sum_matrix(5))
print('\n')

def create_sum_matrix2(n):
    # Create an n x n array using broadcasting
    m = np.ones((n, n), dtype=int)
    # Set the first row to 0, 1, ..., n-1
    m[0, :] = np.arange(n)
    return np.cumsum(m, axis=0)

print('Using cumsum:')
print(create_sum_matrix2(5))
```

# Problem: 8
Evaluate $cos$ and $sin$ on the interval $[0,1]$  with 1000 points and then stack the results into a tall array with rows being the $(cos(x),sin(x))$ entries.

```python
x = np.linspace(0, 1, 1000)

cos_x = np.cos(x)
sin_x = np.sin(x)

print('cos_x shape: ', cos_x.shape)  # Output: (1000,)
print('sin_x shape: ', sin_x.shape)  # Output: (1000,)

# Stack into a tall array with rows as (cos(x), sin(x))
stacked_array = np.column_stack((cos_x, sin_x))

print('column_stack shape: ', stacked_array.shape)  # Output: (1000,2)

#Using vstack
stacked_array2 = np.vstack((np.cos(x), np.sin(x))).T
print('vstack shape: ', stacked_array2.shape)  # Output: (1000,2)


#Using hstack
stacked_array3 = np.hstack((np.cos(x), np.sin(x)))
print('hstack shape: ', stacked_array3.shape)  # Output: (2000,)
```

# Problem: 9

Create a random $3×5$ array using the `np.random.rand(3 ,5)` function and compute: the sum of all the entries, the sum of the rows, and the sum of the columns.

```python
arr = np.random.rand(3, 5)

# the sum of all the entries
total_sum = np.sum(arr)

# the sum of the rows
row_sum = np.sum(arr, axis = 0)

# the sum of the columns
col_sum = np.sum(arr, axis = 1)

print("Array:\n", arr)
print("\nSum of all entries:", total_sum)
print("Sum of the rows:", row_sum)
print("Sum of the columns:", col_sum)
```

## Problem: 10

Create a random $5×5$ array using the function `np.random.rand(5, 5)`. Then sort the rows according to the second column.

```python
arr = np.random.randint(0, 10, size=(5, 5))
print("Array:\n", arr)

sorted_arr = np.sort(arr, axis=0)
print("\nSorted Arrary:\n", sorted_arr)

print('\nArg sort order', arr[:, 2].argsort())

sorted_col_2 = arr[arr[:, 1].argsort()]
print("\nSorted acc to 2nd Col:\n", sorted_col_2)
```
