#### 12. What States have the highest estimated percent change between the 2010 Census and the 2017 Population Estimate?

**NOTE:** This will take several lines of code, as it requires a recalculation of Percent Change.

```python
state_percent_change = pop.groupby('State')[['2010Census', '2017PopEstimate']].sum()

state_percent_change.head()
```

```python
state_percent_change['PercentChange'] = (((state_percent_change['2017PopEstimate'] - state_percent_change['2010Census']) / state_percent_change['2010Census'] ) * 100).round(2)

state_percent_change.sort_values(by = 'PercentChange', ascending=False).head()
```

```python
state_percent_change.sort_values(by = 'PercentChange', ascending=False).index[0]
```
